## Perform Exploring Data - EDA Technique

There are two common techniques to perform EDA: exploring and visualizing. In the exploration phase, we filter, group, and sort our data to summarize it and build a clearer understanding of what we are working with.

Keep in mind that the goal of EDA is to examine our data from multiple angles; uncovering patterns, relationships, and insights along the way. For example, we can explore how our data is distributed between normal and legendary Pokémon

### Filtering Pokemon by stats, type and legendary record

Legendary Pokémon tend to have higher stats, so we will filter Pokémon with a total stat sum greater than 500 and check which ones are legendary.

Additionally, we'll review the types for each Pokémon. Some legendary Pokémon tend to have Psychic or Dragon as their primary type, and often feature dual typing.

```python
# Filtering Pokemon whose stats are greater than 500
stats_mask = df_pokemon['Total Stats'] > 500

filtered_pokemon = df_pokemon[stats_mask]
```

```python
# Checking out legendary pokemons and its stats
filtered_pokemon[filtered_pokemon['is_legendary'] == 1]
```

```python
# Filter by pokemon whose second type is null
filtered_pokemon[filtered_pokemon['type2'].isna()]
```

```python
type1_mask = (filtered_pokemon['type1'] == 'psychic') | (filtered_pokemon['type1'] == 'dragon')

filtered_pokemon[type1_mask]
```

Based on the filtering operations performed previously, we found that not all legendary Pokémon have a total stat sum greater than 500. Some of the filtered Pokémon are actually final evolutions, such as Blastoise and Alakazam.

Additionally, there is a total of 68 legendary Pokémon, which we'll need to distribute across our sets for ML training. We also confirmed that most legendaries have Psychic or Dragon as their primary type, a useful insight that will support our classifier.
