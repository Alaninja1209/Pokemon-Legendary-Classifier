## Perform Exploring Data - EDA Technique

There are two common techniques to perform EDA: exploring and visualizing. In the exploration phase, we filter, group, and sort our data to summarize it and build a clearer understanding of what we are working with.

Keep in mind that the goal of EDA is to examine our data from multiple angles; uncovering patterns, relationships, and insights along the way. For example, we can explore how our data is distributed between normal and legendary Pokémon

```python
# Getting an statistical overview from our dataset
df_pokemon.describe()
```

### Filtering Pokemon by stats, type and legendary record

Legendary Pokémon tend to have higher stats, so we will filter Pokémon with a total stat sum greater than 500 and check which ones are legendary.

Additionally, we'll review the types for each Pokémon. Some legendary Pokémon tend to have Psychic or Dragon as their primary type, and often feature dual typing.

```python
# Filtering Pokemon whose stats are greater than 500
stats_mask = df_pokemon['Total Stats'] > 500

filtered_pokemon = df_pokemon[stats_mask]
```

```python
# Checking out legendary pokemons and its stats
filtered_pokemon[filtered_pokemon['is_legendary'] == 1]
```

```python
# Filter by pokemon whose second type is null
filtered_pokemon[filtered_pokemon['type2'].isna()]
```

```python
type1_mask = (filtered_pokemon['type1'] == 'psychic') | (filtered_pokemon['type1'] == 'dragon')

filtered_pokemon[type1_mask]
```

Based on the filtering operations performed previously, we found that not all legendary Pokémon have a total stat sum greater than 500. Some of the filtered Pokémon are actually final evolutions, such as Blastoise and Alakazam.

Additionally, there is a total of 68 legendary Pokémon, which we'll need to distribute across our sets for ML training. We also confirmed that most legendaries have Psychic or Dragon as their primary type, a useful insight that will support our classifier.

### Sorting Pokemon by each stat

Now we'll sort our data based on Pokémon stats to get an overall view of which Pokémon have the highest or lowest values in a specific stat. For example, legendary Pokémon tend to have lower Speed but higher HP compared to regular ones.

```python
# Sorting pokemon by Total Stats on descending order
filtered_pokemon.sort_values('Total Stats', ascending = False)
```

```python
# Sorting HP in ascending order
filtered_pokemon.sort_values('hp', ascending = True)
```

```python
# Sorting by speed
filtered_pokemon.sort_values('speed', ascending = True)
```

```python
# Sorting by attack
filtered_pokemon.sort_values('attack', ascending = False)
```

```python
filtered_pokemon.sort_values('sp_attack', ascending = False)
```

As we can see from the previous sorting operations, legendary Pokémon tend to have a higher combined stat total. Regarding each specific stat, there is significant variation. For example:

- Speed: Pokémon at their final stage of evolution (Sceptile or Alakazam) have similar Speed stats to Pheromosa, which is considered one of the fastest Ultra Beasts.
- HP: Pokémon with no evolution line or at their final evolution stage dominate this stat. Blissey, Snorlax, and Slaking are not legendary Pokémon yet have the highest HP concentration — showing a high diversity in this field.
- Sp. Attack: Most legendaries have higher Special Attack, with only a few final evolutions ranking at the top.

Based on these insights, the Total Stats column could be a strong feature for our classifier.

**Note**: For our Legendary Classifier, we will consider **Ultra Beasts** as Legendary Pokémon.

## Perform Visualization Data - EDA Technique

After finishing our exploratory phase, we will proceed to the visualization phase by plotting different graphs, such as histograms and scatterplots, to find patterns, trends, and anomalies that we could investigate further.

### Categorical Charts - First and Second type distribution

First, we plot the distributions of both primary and secondary types for Legendary Pokémon. This allows us to validate our hypothesis that most Legendary Pokémon carry Dragon or Psychic as their primary type, while also providing a clearer understanding of how secondary types are distributed.

```python
# Creating histogram based on First Pokemon Type for legendaries
sns.histplot(y = df_legendary_pkmn['type1'], bins = 18, color = 'red');
```

```python
# Creating histogram based on Second Pokemon Type for legendaries
sns.histplot(y = df_legendary_pkmn['type2'], bins = 18, color = 'green');
```

As we can see, the hypothesis that most Legendary Pokémon have Dragon or Psychic as their primary type holds true, providing an important insight to consider when analyzing feature correlations.

Conversely, when examining secondary types, the distribution shifts significantly; the majority of Legendary Pokémon tend to have Fighting, Fairy, or Flying as their secondary attribute.

### Feature Relationships: Pairplot Analysis

As a next step, we will generate a pairplot matrix to visualize the pairwise relationships and distributions across our numerical base stats. This visualization is crucial for our Legendary Classifier pipeline, as it allows us to:

- **Identify Multi-Collinearity:** Spot highly correlated features (e.g., Base Stat Total vs. Individual Stats) that might introduce redundancy into our model.

- **Detect Outliers and Anomalies:** Isolate extreme data points or skewed distributions that could negatively impact distance-based or linear machine learning algorithms.

- **Assess Feature Separability:** Observe whether Legendary and non-Legendary Pokémon form distinct clusters within the scatterplots, indicating strong predictive power for specific feature combinations.

**Note:** We will use the `legendary` dataframe to explore distributions and relationships. The complete dataframe will then be applied to evaluate correlations between columns.

```python
# Plotting our pairplot
sns.pairplot(df_legendary_pkmn);
```

After plotting the pairwise relationships, several insights emerge:

- Left-Skewed Data (Total Stats): Most of the total stat values are distributed to the left, with two clear outliers separated from the rest. This pattern is expected since Legendary Pokémon typically have higher totals, clustering around ~600 BST.

- Right-Skewed Data (HP and Pokédex Number):  
  - **HP:** The distribution is positively skewed, with a notable cluster between 50–100 HP. This range is uncommon for Legendaries and highlights a subgroup worth further analysis.  
  - **Pokédex Number:** The skew is benign, reflecting the sequential addition of Pokémon across generations.

- Discrete Values (`is_legendary`): The target variable shows a square-shaped distribution due to its binary nature, constrained to values of `0` (Non-Legendary) and `1` (Legendary).
